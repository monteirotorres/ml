# Tipos de aprendizagem

**Objetivo:** usar os mesmos dados para formular três problemas — regressão, classificação e clustering — mostrando que o **tipo** vem da pergunta, não dos dados.

```python
# bibliotecas base
import numpy as np
import pandas as pd

# Plotly para os gráficos (interativos e leves no Colab)
import plotly.graph_objects as go
import plotly.express as px
import plotly.io as pio
pio.templates.default = "simple_white"

# paleta do curso (a mesma do site)
AZUL, VERMELHO, VERDE = "#3266ad", "#c0392b", "#1a7a4a"
TINTA, SUAVE = "#1c1e15", "#6b7050"

# reprodutibilidade: uma única semente para tudo que é aleatório
SEMENTE = 42
np.random.seed(SEMENTE)
```

## Dados simulados

Pacientes com idade, IMC e um marcador contínuo.

```python
n = 200
idade = np.random.uniform(20, 80, n)
imc = np.random.normal(26, 4, n)
marcador = 0.5 * idade + 1.2 * imc + np.random.normal(0, 5, n)
doente = (marcador > np.median(marcador)).astype(int)
df = pd.DataFrame({"idade": idade, "imc": imc, "marcador": marcador, "doente": doente})
df.head()
```

## 1. Regressão — prever o marcador (contínuo)

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

reg = LinearRegression().fit(df[["idade", "imc"]], df["marcador"])
print("R2:", round(r2_score(df["marcador"], reg.predict(df[["idade","imc"]])), 3))
```

## 2. Classificação — prever doente (categórico)

```python
from sklearn.linear_model import LogisticRegression

clf = LogisticRegression().fit(df[["idade", "imc"]], df["doente"])
print("acuracia:", round(clf.score(df[["idade", "imc"]], df["doente"]), 3))
```

## 3. Clustering — descobrir grupos (sem rótulo)

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

Xk = StandardScaler().fit_transform(df[["idade", "imc"]])
grupos = KMeans(n_clusters=2, n_init=10, random_state=SEMENTE).fit_predict(Xk)

figura = go.Figure(go.Scatter(x=df["idade"], y=df["imc"], mode="markers",
                              marker=dict(color=grupos, colorscale="Bluered", size=7)))
figura.update_layout(title="k-means (k=2) sobre idade e IMC",
                     xaxis_title="idade", yaxis_title="imc", height=380,
                     showlegend=False, margin=dict(l=10, r=10, t=50, b=10))
figura.show()
```

```python
# @title Solução
from sklearn.metrics import adjusted_rand_score
print("ARI grupos vs doente:", round(adjusted_rand_score(df["doente"], grupos), 3))
# Regressao: prever o numero exato de internacoes (contagem).
# Classificacao: prever a categoria "0", "1-2", "3+" internacoes.
```
