# Tipos de aprendizagem

**Objetivo:** usar um mesmo conjunto de dados para formular três problemas diferentes — regressão, classificação e clustering — mostrando que o **tipo** vem da pergunta, não dos dados.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Paleta do curso (idêntica ao site)
INK, PAPER = "#1a1a17", "#fcfdf6"
BLUE, RED, GREEN, MUTED = "#3266ad", "#c0392b", "#1a7a4a", "#64684f"

plt.rcParams.update({
    "font.family": "serif", "font.size": 12,
    "figure.facecolor": PAPER, "axes.facecolor": PAPER,
    "axes.edgecolor": "#b3bd95", "axes.grid": True,
    "grid.color": "#dde3c9", "grid.linewidth": 0.7,
    "axes.labelcolor": INK, "text.color": INK,
    "xtick.color": MUTED, "ytick.color": MUTED,
})
rng = np.random.default_rng(42)
```

## Dados simulados

Geramos pacientes com idade, IMC e um marcador contínuo.

```python
n = 200
idade = rng.uniform(20, 80, n)
imc = rng.normal(26, 4, n)
marcador = 0.5 * idade + 1.2 * imc + rng.normal(0, 5, n)
doente = (marcador > np.median(marcador)).astype(int)
df = pd.DataFrame({'idade': idade, 'imc': imc,
                   'marcador': marcador, 'doente': doente})
df.head()
```

## 1. Regressão — prever o marcador (contínuo)

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

Xr, yr = df[['idade', 'imc']], df['marcador']
reg = LinearRegression().fit(Xr, yr)
print('R²:', round(r2_score(yr, reg.predict(Xr)), 3))
```

## 2. Classificação — prever doente (categórico)

```python
from sklearn.linear_model import LogisticRegression

Xc, yc = df[['idade', 'imc']], df['doente']
clf = LogisticRegression().fit(Xc, yc)
print('Acurácia:', round(clf.score(Xc, yc), 3))
```

## 3. Clustering — descobrir grupos (sem rótulo)

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

Xk = StandardScaler().fit_transform(df[['idade', 'imc']])
grupos = KMeans(n_clusters=2, n_init=10, random_state=42).fit_predict(Xk)
plt.scatter(df['idade'], df['imc'], c=grupos, cmap='coolwarm', s=18)
plt.xlabel('idade'); plt.ylabel('imc'); plt.title('k-means (k=2)')
plt.show()
```
